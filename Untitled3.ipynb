```python
import threading
import time
from datetime import datetime
from copy import deepcopy
from statistics import mean

# ============================================================
# 1. CONFIGURAÇÕES
# ============================================================

NUM_NOS = 5
QUORUM = (NUM_NOS // 2) + 1

INTERVALO_HEARTBEAT = 0.10
TIMEOUT_HEARTBEAT = 0.50
INTERVALO_MONITORAMENTO = 0.02

SALDO_INICIAL = 1000.0

print_lock = threading.Lock()


def log(origem, mensagem):
    with print_lock:
        horario = datetime.now().strftime("%H:%M:%S.%f")[:-3]
        print(f"{horario} [{origem}] {mensagem}", flush=True)


# ============================================================
# 2. CLASSE DOS NÓS
# ============================================================

class No:

    def __init__(self, no_id):

        self.id = no_id

        self.ativo = True
        self.rede_disponivel = True
        self.suspeito = False

        self.termo = 0
        self.votou_em = None

        self.log = []
        self.commit_index = 0

        self.ultimo_heartbeat = time.monotonic()
        self.instante_falha = None
        self.tempo_deteccao = None

    def entradas_confirmadas(self):
        return deepcopy(self.log[:self.commit_index])

    def calcular_saldo(self):

        saldo = SALDO_INICIAL

        for entrada in self.entradas_confirmadas():
            saldo -= entrada["valor"]

        return saldo

    def versao(self):
        return self.commit_index

    def ids_transacoes(self):

        return [
            entrada["id"]
            for entrada in self.entradas_confirmadas()
        ]

    def estado(self):

        return (
            round(self.calcular_saldo(), 2),
            self.commit_index,
            tuple(self.ids_transacoes())
        )

    def sincronizar(self, origem):

        self.log = deepcopy(origem.log)
        self.commit_index = origem.commit_index
        self.termo = origem.termo
        self.votou_em = None

        self.ultimo_heartbeat = time.monotonic()
        self.suspeito = False


# ============================================================
# 3. CLASSE DO SISTEMA DISTRIBUÍDO
# ============================================================

class SistemaDistribuido:

    def __init__(self, algoritmo):

        self.algoritmo = algoritmo

        self.nos = [
            No(i)
            for i in range(1, NUM_NOS + 1)
        ]

        self.primario_id = 1
        self.termo_atual = 0

        self.lock = threading.RLock()

        self.parar_eventos = threading.Event()
        self.threads = []

        self.executando = False

        self.metricas = {
            "transacoes_aprovadas": 0,
            "transacoes_rejeitadas": 0,
            "leituras_realizadas": 0,
            "leituras_rejeitadas": 0,
            "falhas_detectadas": 0,
            "eleicoes_realizadas": 0,
            "eleicoes_sem_quorum": 0,
            "recuperacoes_realizadas": 0,
            "particoes_simuladas": 0,
            "tempos_deteccao": [],
            "tempos_eleicao": [],
            "tempos_recuperacao": [],
            "tempos_escrita": [],
            "tempos_leitura": []
        }

    # --------------------------------------------------------
    # 3.1 ACESSO AOS NÓS
    # --------------------------------------------------------

    def obter_no(self, no_id):
        return self.nos[no_id - 1]

    def primario(self):

        if self.primario_id is None:
            return None

        no = self.obter_no(self.primario_id)

        if no.ativo and no.rede_disponivel:
            return no

        return None

    def nos_disponiveis(self):

        return [
            no for no in self.nos
            if no.ativo and no.rede_disponivel
        ]

    def nos_ativos(self):

        return [
            no for no in self.nos
            if no.ativo
        ]

    # --------------------------------------------------------
    # 4. HEARTBEAT INDIVIDUAL
    # --------------------------------------------------------

    def heartbeat(self):

        while not self.parar_eventos.is_set():

            with self.lock:

                lider = self.primario()

                if lider is not None:

                    agora = time.monotonic()

                    # O primário envia heartbeat individual
                    # para cada secundário disponível.

                    for no in self.nos:

                        if no.id == lider.id:
                            continue

                        if no.ativo and no.rede_disponivel:

                            no.ultimo_heartbeat = agora
                            no.suspeito = False

                            log(
                                "HEARTBEAT",
                                f"Primário {lider.id} -> "
                                f"Nó {no.id}: ACK."
                            )

            self.parar_eventos.wait(INTERVALO_HEARTBEAT)

    # --------------------------------------------------------
    # 5. VOTAÇÃO PARA ELEIÇÃO
    # --------------------------------------------------------

    def solicitar_voto(self, candidato, eleitor):

        if not eleitor.ativo or not eleitor.rede_disponivel:
            return False

        if candidato.termo < eleitor.termo:
            return False

        log(
            "VOTAÇÃO",
            f"Nó {eleitor.id} avaliando candidato {candidato.id}."
        )

        log_atualizado_candidato = (
            candidato.commit_index,
            len(candidato.log)
        )

        log_atualizado_eleitor = (
            eleitor.commit_index,
            len(eleitor.log)
        )

        if log_atualizado_candidato < log_atualizado_eleitor:
            return False

        if eleitor.termo < candidato.termo:

            eleitor.termo = candidato.termo
            eleitor.votou_em = None

        if eleitor.votou_em is None:

            eleitor.votou_em = candidato.id

            log(
                "VOTAÇÃO",
                f"Nó {eleitor.id} votou no candidato {candidato.id}."
            )

            return True

        return eleitor.votou_em == candidato.id

    # --------------------------------------------------------
    # 6. ELEIÇÃO DE UM NOVO PRIMÁRIO
    # --------------------------------------------------------

    def eleger_primario(self):

        inicio = time.monotonic()

        with self.lock:

            candidatos = self.nos_disponiveis()

            if len(candidatos) < QUORUM:

                self.primario_id = None

                self.metricas["eleicoes_sem_quorum"] += 1

                log(
                    "ELEIÇÃO",
                    "Quórum insuficiente. Não existe novo primário."
                )

                return False

            self.termo_atual += 1

            # Escolhe o nó com log mais atualizado.
            # O identificador desempata a eleição.

            candidatos.sort(
                key=lambda no: (
                    -no.commit_index,
                    -len(no.log),
                    no.id
                )
            )

            eleito = None

            for candidato in candidatos:

                candidato.termo = self.termo_atual
                candidato.votou_em = candidato.id

                votos = 1

                log(
                    "ELEIÇÃO",
                    f"Nó {candidato.id} iniciou candidatura "
                    f"no termo {self.termo_atual}."
                )

                for eleitor in candidatos:

                    if eleitor.id == candidato.id:
                        continue

                    if self.solicitar_voto(candidato, eleitor):
                        votos += 1

                log(
                    "ELEIÇÃO",
                    f"Candidato {candidato.id}: "
                    f"{votos} votos de {NUM_NOS}."
                )

                if votos >= QUORUM:
                    eleito = candidato
                    break

            if eleito is None:

                self.primario_id = None

                self.metricas["eleicoes_sem_quorum"] += 1

                log(
                    "ELEIÇÃO",
                    "Nenhum candidato conseguiu maioria."
                )

                return False

            self.primario_id = eleito.id

            # Atualiza o termo dos nós participantes.

            for no in candidatos:
                no.termo = self.termo_atual

            self.metricas["eleicoes_realizadas"] += 1

            tempo = time.monotonic() - inicio

            self.metricas["tempos_eleicao"].append(tempo)

            log(
                "ELEIÇÃO",
                f"NOVO PRIMÁRIO: NÓ {eleito.id}. "
                f"Termo={self.termo_atual}."
            )

            log(
                "ELEIÇÃO",
                f"Tempo de eleição: {tempo:.6f} segundos."
            )

            return True

    # --------------------------------------------------------
    # 7. DETECÇÃO INDIVIDUAL DE FALHAS
    # --------------------------------------------------------

    def monitorar_falhas(self):

        while not self.parar_eventos.is_set():

            time.sleep(INTERVALO_MONITORAMENTO)

            with self.lock:

                agora = time.monotonic()

                # Detecta ausência de heartbeat individual.

                for no in self.nos:

                    if not no.ativo or not no.rede_disponivel:
                        continue

                    if no.id == self.primario_id:
                        continue

                    atraso = agora - no.ultimo_heartbeat

                    if atraso >= TIMEOUT_HEARTBEAT:

                        if not no.suspeito:

                            no.suspeito = True

                            log(
                                "MONITOR",
                                f"Nó {no.id} não respondeu ao heartbeat. "
                                f"Estado: SUSPEITO."
                            )

                # Detecta a perda do primário.
                # A eleição só ocorre após o timeout.

                if self.primario_id is not None:

                    lider = self.obter_no(self.primario_id)

                    if not lider.ativo or not lider.rede_disponivel:

                        seguidores = [
                            no for no in self.nos
                            if no.ativo
                            and no.rede_disponivel
                            and no.id != lider.id
                        ]

                        todos_sem_heartbeat = (
                            len(seguidores) >= QUORUM - 1
                            and all(
                                agora - no.ultimo_heartbeat
                                >= TIMEOUT_HEARTBEAT
                                for no in seguidores
                            )
                        )

                        if todos_sem_heartbeat:

                            if lider.tempo_deteccao is None:

                                lider.tempo_deteccao = (
                                    agora - lider.instante_falha
                                    if lider.instante_falha is not None
                                    else TIMEOUT_HEARTBEAT
                                )

                                self.metricas[
                                    "falhas_detectadas"
                                ] += 1

                                self.metricas[
                                    "tempos_deteccao"
                                ].append(lider.tempo_deteccao)

                                log(
                                    "MONITOR",
                                    f"Falha do primário {lider.id} "
                                    f"detectada após "
                                    f"{lider.tempo_deteccao:.4f} segundos."
                                )

                            self.eleger_primario()

            # Evita execução excessiva do monitor.
            self.parar_eventos.wait(INTERVALO_MONITORAMENTO)

    # --------------------------------------------------------
    # 8. INICIALIZAÇÃO
    # --------------------------------------------------------

    def iniciar(self):

        if self.executando:
            return

        self.executando = True
        self.parar_eventos.clear()

        self.threads = [

            threading.Thread(
                target=self.heartbeat,
                daemon=True
            ),

            threading.Thread(
                target=self.monitorar_falhas,
                daemon=True
            )
        ]

        for thread in self.threads:
            thread.start()

        log("SISTEMA", f"Implementação: {self.algoritmo}.")
        log("SISTEMA", f"Primário inicial: NÓ {self.primario_id}.")
        log("SISTEMA", f"Quórum: {QUORUM} de {NUM_NOS} nós.")

    def parar(self):

        self.parar_eventos.set()

        for thread in self.threads:
            thread.join(timeout=1.0)

        self.executando = False

        log("SISTEMA", "Monitoramento encerrado.")

    # --------------------------------------------------------
    # 9. PREPARAÇÃO E REPLICAÇÃO
    # --------------------------------------------------------

    def preparar_transacao(self, no, transacao):

        if not no.ativo or not no.rede_disponivel:
            return False

        if no.commit_index != self.primario().commit_index:
            return False

        if transacao["id"] in no.ids_transacoes():
            return False

        if transacao["valor"] <= 0:
            return False

        if no.calcular_saldo() < transacao["valor"]:
            return False

        return True

    def escrever(self, transacao):

        inicio = time.monotonic()

        with self.lock:

            lider = self.primario()

            if lider is None:

                log(
                    "ESCRITA",
                    "Operação rejeitada: primário indisponível."
                )

                self.metricas["transacoes_rejeitadas"] += 1
                return False

            participantes = self.nos_disponiveis()

            if len(participantes) < QUORUM:

                log(
                    "ESCRITA",
                    "Operação rejeitada: quórum insuficiente."
                )

                self.metricas["transacoes_rejeitadas"] += 1
                return False

            if transacao["id"] in lider.ids_transacoes():

                log(
                    "ESCRITA",
                    f"Transação duplicada: {transacao['id']}."
                )

                self.metricas["transacoes_rejeitadas"] += 1
                return False

            if transacao["valor"] <= 0:

                log("ESCRITA", "Valor inválido.")

                self.metricas["transacoes_rejeitadas"] += 1
                return False

            if lider.calcular_saldo() < transacao["valor"]:

                log("ESCRITA", "Saldo insuficiente.")

                self.metricas["transacoes_rejeitadas"] += 1
                return False

            if self.algoritmo == "PRIMARIO-SECUNDARIO":

                # FASE 1: PREPARE

                votos = []

                for no in participantes:

                    if self.preparar_transacao(no, transacao):
                        votos.append(no)

                log(
                    "PREPARE",
                    f"Votos positivos: {len(votos)}. "
                    f"Quórum necessário: {QUORUM}."
                )

                if len(votos) < QUORUM:

                    log("ESCRITA", "ABORT: quórum não alcançado.")

                    self.metricas["transacoes_rejeitadas"] += 1
                    return False

                # FASE 2: COMMIT

                entrada = deepcopy(transacao)
                entrada["termo"] = self.termo_atual

                for no in votos:

                    no.log.append(deepcopy(entrada))
                    no.commit_index += 1

            else:

                # RAFT SIMPLIFICADO
                # AppendEntries + confirmação por maioria.

                entrada = deepcopy(transacao)
                entrada["termo"] = self.termo_atual

                anexados = []

                for no in participantes:

                    if no.commit_index != lider.commit_index:
                        continue

                    if no.log != lider.log:
                        continue

                    no.log.append(deepcopy(entrada))
                    anexados.append(no)

                    log(
                        "APPEND ENTRIES",
                        f"Entrada {transacao['id']} enviada ao nó {no.id}."
                    )

                if len(anexados) < QUORUM:

                    log(
                        "RAFT",
                        "Quórum de replicação não alcançado."
                    )

                    self.metricas["transacoes_rejeitadas"] += 1
                    return False

                # Confirmação da entrada nos nós que a receberam.

                for no in anexados:
                    no.commit_index += 1

            log(
                "COMMIT",
                f"Transação {transacao['id']} confirmada."
            )

            log(
                "COMMIT",
                f"Saldo do primário: R$ {lider.calcular_saldo():.2f}. "
                f"Versão: {lider.versao()}."
            )

            self.metricas["transacoes_aprovadas"] += 1

            tempo = time.monotonic() - inicio

            self.metricas["tempos_escrita"].append(tempo)

            return True

    # --------------------------------------------------------
    # 10. LEITURA COM QUÓRUM
    # --------------------------------------------------------

    def ler_saldo(self):

        inicio = time.monotonic()

        with self.lock:

            respostas = self.nos_disponiveis()

            if len(respostas) < QUORUM:

                log(
                    "LEITURA",
                    "Leitura rejeitada: quórum insuficiente."
                )

                self.metricas["leituras_rejeitadas"] += 1
                return None

            # Agrupa respostas pelo índice de commit.

            versoes = {}

            for no in respostas:

                chave = (
                    no.commit_index,
                    no.estado()
                )

                versoes.setdefault(chave, []).append(no)

            # Exige maioria de respostas com o mesmo estado.

            grupos_validos = [
                (chave, grupo)
                for chave, grupo in versoes.items()
                if len(grupo) >= QUORUM
            ]

            if not grupos_validos:

                log(
                    "LEITURA",
                    "Conflito detectado: não existe maioria "
                    "com estado idêntico."
                )

                self.metricas["leituras_rejeitadas"] += 1
                return None

            chave, grupo = max(
                grupos_validos,
                key=lambda item: item[0][0]
            )

            fonte = grupo[0]

            # Read repair: corrige réplicas desatualizadas.

            for no in respostas:

                if no.commit_index < fonte.commit_index:

                    no.sincronizar(fonte)

                    log(
                        "READ REPAIR",
                        f"Nó {no.id} sincronizado para versão "
                        f"{fonte.commit_index}."
                    )

            saldo = fonte.calcular_saldo()

            self.metricas["leituras_realizadas"] += 1

            self.metricas["tempos_leitura"].append(
                time.monotonic() - inicio
            )

            log(
                "LEITURA",
                f"Saldo=R$ {saldo:.2f}; "
                f"versão={fonte.versao()}; "
                f"respostas={len(respostas)}."
            )

            return saldo

    # --------------------------------------------------------
    # 11. SIMULAÇÃO DE FALHAS
    # --------------------------------------------------------

    def falhar_no(self, no_id):

        with self.lock:

            no = self.obter_no(no_id)

            no.ativo = False
            no.instante_falha = time.monotonic()
            no.tempo_deteccao = None

            log("FALHA", f"NÓ {no_id} indisponível.")

    # --------------------------------------------------------
    # 12. PARTIÇÃO DE REDE
    # --------------------------------------------------------

    def particionar_no(self, no_id):

        with self.lock:

            no = self.obter_no(no_id)

            no.rede_disponivel = False
            no.suspeito = False

            self.metricas["particoes_simuladas"] += 1

            log(
                "REDE",
                f"Partição simulada no NÓ {no_id}."
            )

    def restaurar_rede(self, no_id):

        with self.lock:

            no = self.obter_no(no_id)

            no.rede_disponivel = True

            log(
                "REDE",
                f"Conectividade restaurada no NÓ {no_id}."
            )

            return self.recuperar_no(no_id)

    # --------------------------------------------------------
    # 13. RECUPERAÇÃO DE RÉPLICAS
    # --------------------------------------------------------

    def recuperar_no(self, no_id):

        inicio = time.monotonic()

        with self.lock:

            no = self.obter_no(no_id)

            lider = self.primario()

            if lider is None:

                log(
                    "RECUPERAÇÃO",
                    "Não existe primário disponível."
                )

                return False

            fontes = [
                r for r in self.nos_disponiveis()
                if r.id != no_id
            ]

            if not fontes:

                log(
                    "RECUPERAÇÃO",
                    "Nenhuma fonte de sincronização disponível."
                )

                return False

            # O primário é a fonte preferencial.
            # Se o destino for o primário, utiliza a réplica
            # disponível com maior índice de commit.

            if lider.id != no_id:
                origem = lider
            else:
                origem = max(
                    fontes,
                    key=lambda r: r.commit_index
                )

            # Confere se as fontes possuem o mesmo estado
            # confirmado antes da sincronização.

            estados_fontes = {
                (
                    r.commit_index,
                    tuple(
                        (e["id"], e["valor"])
                        for e in r.entradas_confirmadas()
                    )
                )
                for r in fontes
            }

            if len(estados_fontes) > 1:

                log(
                    "RECUPERAÇÃO",
                    "Conflito entre fontes de sincronização. "
                    "Recuperação interrompida."
                )

                return False

            no.sincronizar(origem)

            no.ativo = True
            no.rede_disponivel = True

            tempo = time.monotonic() - inicio

            self.metricas["recuperacoes_realizadas"] += 1

            self.metricas["tempos_recuperacao"].append(tempo)

            log(
                "RECUPERAÇÃO",
                f"NÓ {no_id} sincronizado com NÓ {origem.id}. "
                f"Versão={no.commit_index}."
            )

            log(
                "RECUPERAÇÃO",
                f"Tempo de recuperação: {tempo:.6f} segundos."
            )

            return True

    # --------------------------------------------------------
    # 14. VERIFICAÇÃO DE CONSISTÊNCIA
    # --------------------------------------------------------

    def verificar_consistencia(self):

        with self.lock:

            ativos = self.nos_disponiveis()

            if not ativos:

                log("CONSISTÊNCIA", "Nenhum nó disponível.")
                return False

            estados = {
                no.estado()
                for no in ativos
            }

            consistente = len(estados) == 1

            log(
                "CONSISTÊNCIA",
                "ESTADOS CONSISTENTES."
                if consistente
                else "DIVERGÊNCIA ENTRE RÉPLICAS."
            )

            for no in self.nos:

                situacao = (
                    "ATIVO"
                    if no.ativo and no.rede_disponivel
                    else "INDISPONÍVEL"
                )

                log(
                    "NÓ",
                    f"{no.id}: {situacao} | "
                    f"Saldo=R$ {no.calcular_saldo():.2f} | "
                    f"Versão={no.versao()} | "
                    f"Termo={no.termo}."
                )

            return consistente

    # --------------------------------------------------------
    # 15. TESTE DE DUPLICIDADE E CONFLITO
    # --------------------------------------------------------

    def testar_duplicidade(self, transacao_id):

        with self.lock:

            lider = self.primario()

            if lider is None:
                return False

            existe = transacao_id in lider.ids_transacoes()

            log(
                "CONFLITO",
                f"Transação {transacao_id}: "
                f"{'duplicidade identificada' if existe else 'não encontrada'}."
            )

            return existe

    # --------------------------------------------------------
    # 16. RELATÓRIO DE MÉTRICAS
    # --------------------------------------------------------

    def relatorio_metricas(self):

        log(
            "MÉTRICAS",
            f"========== {self.algoritmo} =========="
        )

        for chave in [
            "transacoes_aprovadas",
            "transacoes_rejeitadas",
            "leituras_realizadas",
            "leituras_rejeitadas",
            "falhas_detectadas",
            "eleicoes_realizadas",
            "eleicoes_sem_quorum",
            "recuperacoes_realizadas",
            "particoes_simuladas"
        ]:

            log(
                "MÉTRICAS",
                f"{chave}: {self.metricas[chave]}"
            )

        def exibir_media(nome, valores):

            if valores:

                log(
                    "MÉTRICAS",
                    f"{nome}: {mean(valores):.6f} segundos."
                )

            else:

                log(
                    "MÉTRICAS",
                    f"{nome}: sem medições."
                )

        exibir_media(
            "Tempo médio de detecção",
            self.metricas["tempos_deteccao"]
        )

        exibir_media(
            "Tempo médio de eleição",
            self.metricas["tempos_eleicao"]
        )

        exibir_media(
            "Tempo médio de recuperação",
            self.metricas["tempos_recuperacao"]
        )

        exibir_media(
            "Tempo médio de escrita",
            self.metricas["tempos_escrita"]
        )

        exibir_media(
            "Tempo médio de leitura",
            self.metricas["tempos_leitura"]
        )

        log("MÉTRICAS", "========================================")


# ============================================================
# 17. FUNÇÃO AUXILIAR PARA AGUARDAR ELEIÇÃO
# ============================================================

def aguardar_eleicao(sistema, tempo_maximo=3.0):

    limite = time.monotonic() + tempo_maximo

    while time.monotonic() < limite:

        with sistema.lock:

            lider = sistema.primario()

            if lider is not None:
                return lider.id

        time.sleep(0.02)

    return None


# ============================================================
# 18. EXECUÇÃO SISTEMÁTICA DE UMA IMPLEMENTAÇÃO
# ============================================================

def executar_testes(algoritmo):

    print("\n")
    print("=" * 65)
    print("INICIANDO TESTES:", algoritmo)
    print("=" * 65)

    sistema = SistemaDistribuido(algoritmo)

    sistema.iniciar()

    time.sleep(0.30)

    # --------------------------------------------------------
    # TESTE 1 - OPERAÇÃO NORMAL
    # --------------------------------------------------------

    log("TESTE 1", "Cinco nós disponíveis.")

    sistema.escrever({
        "id": "PIX-001",
        "valor": 120.0
    })

    sistema.ler_saldo()
    sistema.verificar_consistencia()

    print("\n")

    # --------------------------------------------------------
    # TESTE 2 - DUAS RÉPLICAS INDISPONÍVEIS
    # --------------------------------------------------------

    log("TESTE 2", "Falha dos nós 4 e 5.")

    sistema.falhar_no(4)
    sistema.falhar_no(5)

    sistema.escrever({
        "id": "PIX-002",
        "valor": 80.0
    })

    sistema.ler_saldo()
    sistema.verificar_consistencia()

    print("\n")

    # --------------------------------------------------------
    # TESTE 3 - RECUPERAÇÃO DAS RÉPLICAS
    # --------------------------------------------------------

    log("TESTE 3", "Recuperação dos nós 4 e 5.")

    sistema.recuperar_no(4)
    sistema.recuperar_no(5)

    sistema.verificar_consistencia()

    print("\n")

    # --------------------------------------------------------
    # TESTE 4 - FALHA DO PRIMÁRIO E ELEIÇÃO
    # --------------------------------------------------------

    log("TESTE 4", "Falha do primário atual.")

    antigo_primario = sistema.primario_id

    inicio = time.monotonic()

    sistema.falhar_no(antigo_primario)

    novo_primario = aguardar_eleicao(sistema)

    tempo_total = time.monotonic() - inicio

    log(
        "TESTE 4",
        f"Novo primário: {novo_primario}."
    )

    log(
        "TESTE 4",
        f"Tempo total de recuperação da liderança: "
        f"{tempo_total:.6f} segundos."
    )

    if novo_primario is not None:

        sistema.escrever({
            "id": "PIX-003",
            "valor": 50.0
        })

        sistema.ler_saldo()

    sistema.verificar_consistencia()

    print("\n")

    # --------------------------------------------------------
    # TESTE 5 - RECUPERAÇÃO DO ANTIGO PRIMÁRIO
    # --------------------------------------------------------

    log("TESTE 5", "Recuperação do antigo primário.")

    sistema.recuperar_no(antigo_primario)

    sistema.verificar_consistencia()

    print("\n")

    # --------------------------------------------------------
    # TESTE 6 - PERDA DE QUÓRUM
    # --------------------------------------------------------

    log("TESTE 6", "Indisponibilidade de três nós.")

    for no_id in [3, 4, 5]:
        sistema.falhar_no(no_id)

    time.sleep(0.15)

    sistema.escrever({
        "id": "PIX-004",
        "valor": 40.0
    })

    sistema.ler_saldo()

    print("\n")

    # --------------------------------------------------------
    # TESTE 7 - RESTAURAÇÃO DO QUÓRUM
    # --------------------------------------------------------

    log("TESTE 7", "Recuperação dos nós 3, 4 e 5.")

    for no_id in [3, 4, 5]:
        sistema.recuperar_no(no_id)

    sistema.verificar_consistencia()

    print("\n")

    # --------------------------------------------------------
    # TESTE 8 - NOVA TRANSAÇÃO
    # --------------------------------------------------------

    log("TESTE 8", "Nova tentativa após recuperação.")

    sistema.escrever({
        "id": "PIX-004",
        "valor": 40.0
    })

    sistema.ler_saldo()
    sistema.verificar_consistencia()

    print("\n")

    # --------------------------------------------------------
    # TESTE 9 - DUPLICIDADE
    # --------------------------------------------------------

    log("TESTE 9", "Verificação de transação duplicada.")

    sistema.testar_duplicidade("PIX-004")

    sistema.escrever({
        "id": "PIX-004",
        "valor": 40.0
    })

    print("\n")

    # --------------------------------------------------------
    # TESTE 10 - PARTIÇÃO DE REDE
    # --------------------------------------------------------

    log("TESTE 10", "Partição de rede no nó 5.")

    sistema.particionar_no(5)

    time.sleep(TIMEOUT_HEARTBEAT + 0.15)

    sistema.escrever({
        "id": "PIX-005",
        "valor": 30.0
    })

    sistema.ler_saldo()

    print("\n")

    # --------------------------------------------------------
    # TESTE 11 - RESTABELECIMENTO DA REDE
    # --------------------------------------------------------

    log("TESTE 11", "Restabelecimento da conectividade.")

    sistema.restaurar_rede(5)

    sistema.verificar_consistencia()

    print("\n")

    # --------------------------------------------------------
    # TESTE 12 - TESTE FINAL
    # --------------------------------------------------------

    log("TESTE 12", "Verificação final do sistema.")

    sistema.escrever({
        "id": "PIX-006",
        "valor": 20.0
    })

    sistema.ler_saldo()
    sistema.verificar_consistencia()

    print("\n")

    # --------------------------------------------------------
    # RELATÓRIO FINAL
    # --------------------------------------------------------

    sistema.relatorio_metricas()

    resultado = {
        "algoritmo": algoritmo,
        "metricas": deepcopy(sistema.metricas),
        "saldo_final": sistema.obter_no(
            sistema.primario_id
        ).calcular_saldo() if sistema.primario_id else None,
        "versao_final": sistema.obter_no(
            sistema.primario_id
        ).versao() if sistema.primario_id else None,
        "consistente": sistema.verificar_consistencia()
    }

    sistema.parar()

    return resultado


# ============================================================
# 19. COMPARAÇÃO DAS DUAS IMPLEMENTAÇÕES
# ============================================================

resultado_pb = executar_testes("PRIMARIO-SECUNDARIO")

resultado_raft = executar_testes("RAFT SIMPLIFICADO")


# ============================================================
# 20. COMPARAÇÃO FINAL DAS MÉTRICAS
# ============================================================

print("\n")
print("=" * 75)
print("COMPARAÇÃO FINAL DAS IMPLEMENTAÇÕES")
print("=" * 75)

resultados = [resultado_pb, resultado_raft]

for resultado in resultados:

    metricas = resultado["metricas"]

    print("\nIMPLEMENTAÇÃO:", resultado["algoritmo"])

    print(
        "Transações aprovadas:",
        metricas["transacoes_aprovadas"]
    )

    print(
        "Transações rejeitadas:",
        metricas["transacoes_rejeitadas"]
    )

    print(
        "Falhas detectadas:",
        metricas["falhas_detectadas"]
    )

    print(
        "Eleições realizadas:",
        metricas["eleicoes_realizadas"]
    )

    print(
        "Recuperações:",
        metricas["recuperacoes_realizadas"]
    )

    print(
        "Saldo final:",
        resultado["saldo_final"]
    )

    print(
        "Versão final:",
        resultado["versao_final"]
    )

    print(
        "Consistência final:",
        resultado["consistente"]
    )

    if metricas["tempos_deteccao"]:

        print(
            "Detecção média:",
            f"{mean(metricas['tempos_deteccao']):.6f} s"
        )

    if metricas["tempos_eleicao"]:

        print(
            "Eleição média:",
            f"{mean(metricas['tempos_eleicao']):.6f} s"
        )

    if metricas["tempos_recuperacao"]:

        print(
            "Recuperação média:",
            f"{mean(metricas['tempos_recuperacao']):.6f} s"
        )

    if metricas["tempos_escrita"]:

        print(
            "Escrita média:",
            f"{mean(metricas['tempos_escrita']):.6f} s"
        )

    if metricas["tempos_leitura"]:

        print(
            "Leitura média:",
            f"{mean(metricas['tempos_leitura']):.6f} s"
        )


# ============================================================
# 21. ANÁLISE EXPERIMENTAL DO TEOREMA CAP
# ============================================================

print("\n")
print("=" * 75)
print("ANÁLISE DO TEOREMA CAP")
print("=" * 75)

print("""
CONSISTÊNCIA (C):
As operações confirmadas são replicadas por maioria.
Leituras exigem quórum de respostas com estado idêntico.

DISPONIBILIDADE (A):
O sistema continua aceitando operações enquanto existe
um primário válido e um quórum de nós disponíveis.

TOLERÂNCIA A PARTIÇÕES (P):
O sistema continua operando no lado da partição que possui
quórum. O lado sem maioria rejeita leituras e escritas.

TRADE-OFF OBSERVADO:
Durante a perda do quórum, o sistema bloqueia operações
para preservar a consistência, sacrificando disponibilidade.

CONCLUSÃO:
A simulação demonstra o comportamento de uma arquitetura
orientada à consistência e tolerância a partições (CP)
durante as condições de falha testadas.
""")

print("=" * 75)
print("PRÁTICA 3 FINALIZADA.")
print("=" * 75)
```
